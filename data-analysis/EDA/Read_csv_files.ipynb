# 14.Converters

```python
def rename(name):
  if name == 'Royal Challengers Bangalore	':
    return "RCB"
  else:
    return name
```

```python
rename('Royal Challengers Bangalore')
```

```python
pd.read_csv('/content/IPL Matches 2008-2020.csv',converters={'team1':rename})
```
