```python
##Feature Engineering battery capacity
# 5000-->4600=first Level
# 44000-->4030=second
# 4030> -->third

```

```python
def transform_battery_capacity(num):
  if (4600.0<num) :
    return 'best'
  elif (4600.0>num) and (4030<num):
    return 'better'
  else:
    return 'good'
```

```python
df['battery_condition']=df['battery_capacity'].apply(transform_battery_capacity)
```

```python
df[['battery_condition','battery_capacity']].sample(6)
```

```python
df['battery_capacity'].value_counts().sort_values(ascending=False)
```

```python
df['across_pixels']=df['resolution'].str.strip().str.split('x').str.get(0)
```

```python
df['rating_condition']=df['rating'].apply(transform_rating)
df[['rating_condition','rating']].sample(4)
```
