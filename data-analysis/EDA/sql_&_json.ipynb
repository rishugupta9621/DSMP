```python
df1=pd.DataFrame()
```

```python
df1
```

```python
for i in range(1,501):
  response=requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=2da4aade0fab2a3dbddda8addef95449&&language=en-US&page={}'.format(i))
  json_data = response.json()
  if 'results' in json_data:
    temp_df=pd.DataFrame(json_data['results'])
    df1=pd.concat([df1, temp_df], ignore_index=True)
  else:
    print(f"Warning: 'results' key not found in response for page {i}. Response: {json_data}")
```
